```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.integrate import odeint

# define the function that returns the derivative
def dxdt(x,t, a,b,c,d,K,n):
    # x = [x_1, x_2]  ==> x[0] == x_1 and x[1] == x_2
    x_1, x_2 = x[0], x[1]
    dx_1 = a*x_2**n/(K+x_2**n) - b*x_1
    dx_2 = c*x_1 - d*x_2/(4+x_2)
    return [dx_1, dx_2]
```

```python
# define the parameters
b=2.8 # mRNA degradation rate
d=2 # protein degradation rate
a=4.5 # maximal mRNA production rate
c=1 # translation rate per mRNA
n=4 # cooperativity
K=2.2**n # half-max concentration

plt.figure(figsize=(8,8))
x_2 = np.linspace(0,20,101)
plt.plot(x_2, a/b*x_2**n/(K+x_2**n))
plt.plot(x_2, d/c*x_2/(4+x_2), c='C1')
plt.ylabel('[mRNA] x_1')
plt.xlabel('[Protein] x_2')
```
